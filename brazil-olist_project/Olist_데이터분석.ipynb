#배송지연에_따른_리뷰점수_회귀모델

```python
df6['order_purchase_timestamp'] = pd.to_datetime(df6['order_purchase_timestamp'])
df6['order_delivered_customer_date'] = pd.to_datetime(df6['order_delivered_customer_date'])
df6['order_estimated_delivery_date'] = pd.to_datetime(df6['order_estimated_delivery_date'])
((df6['order_delivered_customer_date'] - df6['order_purchase_timestamp']).dt.days).mean()
```

```python
# 주문일 ~ 실제 배송일 사이의 일수
df6['delivery_days'] = (df6['order_delivered_customer_date'] - df6['order_purchase_timestamp']).dt.days
df6
# reviews: order_id, review_score
df_merge = pd.merge(df6[['order_id','delivery_days']],
                    df5[['order_id','review_score']],
                   on='order_id', how='inner')
df_merge = df_merge.dropna(subset=['delivery_days','review_score'])
low_reviews = df_merge[df_merge['review_score'] <= 2]
low_reviews['delivery_days'].mean()
```

```python
import statsmodels.api as sm

X = sm.add_constant(df_merge['delivery_days'])   # const 추가는 클린업 이후!
y = df_merge['review_score']

model = sm.OLS(y, X, missing='drop').fit(cov_type='HC1')  # 견고표준오차(HC1) 권장
print(model.summary())
```

#배송 거리(Geolocation) × 리뷰/지연

```python
import pandas as pd
import numpy as np
import geopandas as gpd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

# ================================
# 1. 데이터 로드
# ================================
customers = pd.read_csv("/content/drive/MyDrive/archive/olist_customers_dataset.csv")
sellers = pd.read_csv("/content/drive/MyDrive/archive/olist_sellers_dataset.csv")
geolocation = pd.read_csv("/content/drive/MyDrive/archive/olist_geolocation_dataset.csv")
orders = pd.read_csv("/content/drive/MyDrive/archive/olist_orders_dataset.csv")
reviews = pd.read_csv("/content/drive/MyDrive/archive/olist_order_reviews_dataset.csv")
order_items = pd.read_csv("/content/drive/MyDrive/archive/olist_order_items_dataset.csv")

# ================================
# 2. 위경도 평균값 (ZIP 코드 단위)
# ================================
geo_mean = geolocation.groupby("geolocation_zip_code_prefix")[["geolocation_lat", "geolocation_lng"]].mean().reset_index()

# 고객 + 좌표
customers_geo = customers.merge(
    geo_mean, left_on="customer_zip_code_prefix", right_on="geolocation_zip_code_prefix", how="left"
).rename(columns={"geolocation_lat":"customer_lat", "geolocation_lng":"customer_lng"})

# 셀러 + 좌표
sellers_geo = sellers.merge(
    geo_mean, left_on="seller_zip_code_prefix", right_on="geolocation_zip_code_prefix", how="left"
).rename(columns={"geolocation_lat":"seller_lat", "geolocation_lng":"seller_lng"})

# ================================
# 3. 주문 + 셀러 매핑
# ================================
orders_items_joined = (
    orders.merge(order_items[["order_id","seller_id"]], on="order_id", how="inner")
    .merge(customers_geo, on="customer_id", how="left")
    .merge(sellers_geo[["seller_id","seller_lat","seller_lng"]], on="seller_id", how="left")
)

# ================================
# 4. 리뷰 + 배송 지연 추가
# ================================
orders_items_joined = orders_items_joined.merge(reviews[["order_id","review_score"]], on="order_id", how="left")

orders_items_joined["order_delivered_customer_date"] = pd.to_datetime(orders_items_joined["order_delivered_customer_date"])
orders_items_joined["order_estimated_delivery_date"] = pd.to_datetime(orders_items_joined["order_estimated_delivery_date"])
orders_items_joined["delay_days"] = (
    (orders_items_joined["order_delivered_customer_date"] - orders_items_joined["order_estimated_delivery_date"]).dt.days
)

# ================================
# 5. 거리 계산 (Haversine formula)
# ================================
def haversine(lon1, lat1, lon2, lat2):
    lon1, lat1, lon2, lat2 = map(np.radians, [lon1, lat1, lon2, lat2])
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    r = 6371  # km
    return c * r

orders_items_joined["distance_km"] = haversine(
    orders_items_joined["customer_lng"], orders_items_joined["customer_lat"],
    orders_items_joined["seller_lng"], orders_items_joined["seller_lat"]
)

# ================================
# 6. 지역별 집계 (State 단위)
# ================================
region_summary = orders_items_joined.groupby("customer_state").agg(
    avg_distance_km=("distance_km","mean"),
    avg_delay_days=("delay_days","mean"),
    avg_review_score=("review_score","mean"),
    order_count=("order_id","count")
).reset_index()

# ================================
# 7. 브라질 GeoJSON (State boundaries)
# ================================
url = "https://raw.githubusercontent.com/codeforamerica/click_that_hood/master/public/data/brazil-states.geojson"
brazil_states = gpd.read_file(url)

# customer_state 와 geojson 'sigla' 매칭
geo = brazil_states.merge(region_summary, left_on="sigla", right_on="customer_state", how="left")

# ================================
# 8. 시각화 (정적 Choropleth)
# ================================
fig, ax = plt.subplots(1, 1, figsize=(15, 15))

# Choropleth: 평균 배송 지연일수
geo.plot(
    column="avg_delay_days",
    cmap="YlOrRd",
    linewidth=0.8,
    edgecolor="0.8",
    legend=True,
    legend_kwds={"label": "Average Delivery Delay (days)"},
    ax=ax
)

# 라벨 추가 (일반 텍스트 버전)
for idx, row in geo.iterrows():
    if pd.notnull(row["avg_delay_days"]):
        plt.annotate(
            f"{row['sigla']}\n {row['avg_distance_km']:.0f} km \n{row['avg_delay_days']:.1f} d",
            xy=(row.geometry.centroid.x, row.geometry.centroid.y),
            ha="center", fontsize=8, color="black"
        )


ax.set_title("Brazil - Avg Delivery Delay by State", fontsize=16)
ax.axis("off")
plt.show()
```
